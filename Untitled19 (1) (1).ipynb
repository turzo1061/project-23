```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D, Input
from sklearn.metrics import classification_report, confusion_matrix, roc_curve, auc
from sklearn.preprocessing import label_binarize
from IPython.display import display, Markdown
from time import perf_counter
import matplotlib.cm as cm # For ROC curve colors

def printmd(string):
    display(Markdown(string))


def train_and_evaluate_model(model_name, base_model_constructor, preprocess_input_func):
    printmd(f"## Training and Evaluating {model_name}")

    # 1. Base Model Setup
    base_model = base_model_constructor(
        include_top=False,
        input_shape=(224, 224, 3), # Assuming input shape is consistent
        weights='imagenet'
    )
    base_model.trainable = False

    # 2. Build Classification Head
    inputs = tf.keras.Input(shape=(224, 224, 3))
    x = preprocess_input_func(inputs)
    x = base_model(x, training=False)
    x = tf.keras.layers.GlobalAveragePooling2D()(x)
    outputs = tf.keras.layers.Dense(NUM_CLASSES, activation='softmax')(x)
    model = tf.keras.Model(inputs, outputs)

    # 3. Compile Model
    model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

    # 4. Data Generators with specific preprocess_input
    train_datagen = ImageDataGenerator(
        rotation_range=20,
        width_shift_range=0.2,
        height_shift_range=0.2,
        shear_range=0.2,
        zoom_range=0.2,
        horizontal_flip=True,
        fill_mode='nearest',
        preprocessing_function=preprocess_input_func # Use model-specific preprocess_input
    )
    val_datagen = ImageDataGenerator(preprocessing_function=preprocess_input_func)
    test_datagen = ImageDataGenerator(preprocessing_function=preprocess_input_func)

    # 5. Create new generators
    BATCH_SIZE = 32 # Assuming a default batch size

    # Convert 'filepath' column to string type for all dataframes
    # The global dataframes (train_df, val_df, test_df) are being used here.
    # Ensure they have string filepaths before passing to flow_from_dataframe.
    _train_df = train_df.copy()
    _val_df = val_df.copy()
    _test_df = test_df.copy()

    _train_df['filepath'] = _train_df['filepath'].astype(str)
    _val_df['filepath'] = _val_df['filepath'].astype(str)
    _test_df['filepath'] = _test_df['filepath'].astype(str)

    train_generator = train_datagen.flow_from_dataframe(
        _train_df,
        x_col='filepath',
        y_col='label',
        target_size=(224, 224),
        batch_size=BATCH_SIZE,
        class_mode='categorical',
        seed=SEED
    )
    val_generator = val_datagen.flow_from_dataframe(
        _val_df,
        x_col='filepath',
        y_col='label',
        target_size=(224, 224),
        batch_size=BATCH_SIZE,
        class_mode='categorical',
        seed=SEED
    )
    test_generator = test_datagen.flow_from_dataframe(
        _test_df,
        x_col='filepath',
        y_col='label',
        target_size=(224, 224),
        batch_size=BATCH_SIZE,
        class_mode='categorical',
        seed=SEED,
        shuffle=False # Keep order for accurate evaluation
    )

    # Ensure class indices are consistent across generators
    train_generator.class_indices = labels_map
    val_generator.class_indices = labels_map
    test_generator.class_indices = labels_map


    # 6. Train Model
    start_time = perf_counter()
    history = model.fit(
        train_generator,
        epochs=EPOCHS, # Use the global EPOCHS = 100
        validation_data=val_generator,
        verbose=1
    )
    duration = perf_counter() - start_time
    print(f"{model_name} trained in {duration:.2f} sec")

    # Store history
    hist_df = pd.DataFrame(history.history)
    hist_df.index.name = 'epoch'
    hist_df.to_csv(f'{model_name}_training_history.csv')

    # 7. Evaluate Model
    results = model.evaluate(test_generator, verbose=1)
    test_loss, test_accuracy = results[0], results[1]
    print(f"Test Loss for {model_name}: {test_loss:.4f}")
    print(f"Test Accuracy for {model_name}: {test_accuracy:.4f}")

    # 8. Predict on Test Set
    y_prob = model.predict(test_generator)
    pred_classes = np.argmax(y_prob, axis=1)
    y_true = test_generator.classes # True labels from the generator

    # 9. Generate and Plot
    # Accuracy/Loss curves
    plt.figure(figsize=(12, 6))
    plt.plot(hist_df['accuracy'], label='Training Accuracy')
    plt.plot(hist_df['val_accuracy'], label='Validation Accuracy')
    plt.title(f'{model_name} Training and Validation Accuracy')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy')
    plt.legend()
    plt.grid(True)
    plt.show()

    plt.figure(figsize=(12, 6))
    plt.plot(hist_df['loss'], label='Training Loss')
    plt.plot(hist_df['val_loss'], label='Validation Loss')
    plt.title(f'{model_name} Training and Validation Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    plt.grid(True)
    plt.show()

    # Classification Report
    # Ensure target_names align with integer labels based on labels_map
    report = classification_report(y_true, pred_classes, target_names=list(labels_map.keys()), output_dict=True)
    report_df = pd.DataFrame(report).transpose()
    display(report_df)
    report_df.to_csv(f'{model_name}_classification_report.csv')
    print(f"Classification report saved to {model_name}_classification_report.csv")

    # Confusion Matrix
    cm = confusion_matrix(y_true, pred_classes)
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels_map.keys(), yticklabels=labels_map.keys())
    plt.title(f'{model_name} Confusion Matrix')
    plt.xlabel('Predicted Label')
    plt.ylabel('True Label')
    plt.savefig(f'{model_name}_confusion_matrix.png')
    plt.show()
    pd.DataFrame(cm, index=labels_map.keys(), columns=labels_map.keys()).to_csv(f'{model_name}_confusion_matrix.csv')
    print(f"Confusion matrix saved to {model_name}_confusion_matrix.csv")

    # ROC Curves (for multi-class classification)
    # Binarize y_true
    y_true_bin = label_binarize(y_true, classes=np.arange(NUM_CLASSES))

    # Compute ROC curve and ROC area for each class
    fpr = dict()
    tpr = dict()
    roc_auc = dict()
    for i in range(NUM_CLASSES):
        fpr[i], tpr[i], _ = roc_curve(y_true_bin[:, i], y_prob[:, i])
        roc_auc[i] = auc(fpr[i], tpr[i])

    # Compute micro-average ROC curve and ROC area
    fpr["micro"], tpr["micro"], _ = roc_curve(y_true_bin.ravel(), y_prob.ravel())
    roc_auc["micro"] = auc(fpr["micro"], tpr["micro"])

    # Compute macro-average ROC curve and ROC area
    all_fpr = np.unique(np.concatenate([fpr[i] for i in range(NUM_CLASSES)]))
    mean_tpr = np.zeros_like(all_fpr)
    for i in range(NUM_CLASSES):
        mean_tpr += np.interp(all_fpr, fpr[i], tpr[i])
    mean_tpr /= NUM_CLASSES

    fpr["macro"] = all_fpr
    tpr["macro"] = mean_tpr
    roc_auc["macro"] = auc(fpr["macro"], tpr["macro"])

    # Plot all ROC curves
    plt.figure(figsize=(10, 8))
    plt.plot(fpr["micro"], tpr["micro"],
             label=f'micro-average ROC curve (area = {roc_auc["micro"]:.2f})',
             color='deeppink', linestyle=':', linewidth=4)

    plt.plot(fpr["macro"], tpr["macro"],
             label=f'macro-average ROC curve (area = {roc_auc["macro"]:.2f})',
             color='navy', linestyle=':', linewidth=4)

    colors = cm.get_cmap('jet', NUM_CLASSES)
    for i, class_name in labels_map.items():
        plt.plot(fpr[i], tpr[i], color=colors(i), lw=2,
                 label=f'ROC curve of class {class_name} (area = {roc_auc[i]:.2f})')

    plt.plot([0, 1], [0, 1], 'k--', lw=2)
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title(f'{model_name} Multi-class ROC')
    plt.legend(loc="lower right")
    plt.savefig(f'{model_name}_roc_curve_macro_micro.png')
    plt.show()

    # 10. Store results
    models[model_name] = {
        'model': model,
        'perf': duration,
        'test_accuracy': test_accuracy,
        'test_loss': test_loss,
        'history': hist_df,
        'classification_report': report_df,
        'confusion_matrix': cm,
        'roc_auc': roc_auc
    }
    printmd(f"### {model_name} Training and Evaluation Complete.")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D, Input
from sklearn.metrics import classification_report, confusion_matrix, roc_curve, auc
from sklearn.preprocessing import label_binarize
from IPython.display import display, Markdown
from time import perf_counter
import matplotlib.cm as cm # For ROC curve colors

def printmd(string):
    display(Markdown(string))


def train_and_evaluate_model(model_name, base_model_constructor, preprocess_input_func):
    printmd(f"## Training and Evaluating {model_name}")

    # 1. Base Model Setup
    base_model = base_model_constructor(
        include_top=False,
        input_shape=(224, 224, 3), # Assuming input shape is consistent
        weights='imagenet'
    )
    base_model.trainable = False

    # 2. Build Classification Head
    inputs = tf.keras.Input(shape=(224, 224, 3))
    x = preprocess_input_func(inputs)
    x = base_model(x, training=False)
    x = tf.keras.layers.GlobalAveragePooling2D()(x)
    outputs = tf.keras.layers.Dense(NUM_CLASSES, activation='softmax')(x)
    model = tf.keras.Model(inputs, outputs)

    # 3. Compile Model
    model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

    # 4. Data Generators with specific preprocess_input
    train_datagen = ImageDataGenerator(
        rotation_range=20,
        width_shift_range=0.2,
        height_shift_range=0.2,
        shear_range=0.2,
        zoom_range=0.2,
        horizontal_flip=True,
        fill_mode='nearest',
        preprocessing_function=preprocess_input_func # Use model-specific preprocess_input
    )
    val_datagen = ImageDataGenerator(preprocessing_function=preprocess_input_func)
    test_datagen = ImageDataGenerator(preprocessing_function=preprocess_input_func)

    # 5. Create new generators
    BATCH_SIZE = 32 # Assuming a default batch size

    # Convert 'filepath' column to string type for all dataframes
    # The global dataframes (train_df, val_df, test_df) are being used here.
    # Ensure they have string filepaths before passing to flow_from_dataframe.
    _train_df = train_df.copy()
    _val_df = val_df.copy()
    _test_df = test_df.copy()

    _train_df['filepath'] = _train_df['filepath'].astype(str)
    _val_df['filepath'] = _val_df['filepath'].astype(str)
    _test_df['filepath'] = _test_df['filepath'].astype(str)

    train_generator = train_datagen.flow_from_dataframe(
        _train_df,
        x_col='filepath',
        y_col='label',
        target_size=(224, 224),
        batch_size=BATCH_SIZE,
        class_mode='categorical',
        seed=SEED
    )
    val_generator = val_datagen.flow_from_dataframe(
        _val_df,
        x_col='filepath',
        y_col='label',
        target_size=(224, 224),
        batch_size=BATCH_SIZE,
        class_mode='categorical',
        seed=SEED
    )
    test_generator = test_datagen.flow_from_dataframe(
        _test_df,
        x_col='filepath',
        y_col='label',
        target_size=(224, 224),
        batch_size=BATCH_SIZE,
        class_mode='categorical',
        seed=SEED,
        shuffle=False # Keep order for accurate evaluation
    )

    # Ensure class indices are consistent across generators
    train_generator.class_indices = labels_map
    val_generator.class_indices = labels_map
    test_generator.class_indices = labels_map


    # 6. Train Model
    start_time = perf_counter()
    history = model.fit(
        train_generator,
        epochs=EPOCHS, # Use the global EPOCHS = 100
        validation_data=val_generator,
        verbose=1
    )
    duration = perf_counter() - start_time
    print(f"{model_name} trained in {duration:.2f} sec")

    # Store history
    hist_df = pd.DataFrame(history.history)
    hist_df.index.name = 'epoch'
    hist_df.to_csv(f'{model_name}_training_history.csv')

    # 7. Evaluate Model
    results = model.evaluate(test_generator, verbose=1)
    test_loss, test_accuracy = results[0], results[1]
    print(f"Test Loss for {model_name}: {test_loss:.4f}")
    print(f"Test Accuracy for {model_name}: {test_accuracy:.4f}")

    # 8. Predict on Test Set
    y_prob = model.predict(test_generator)
    pred_classes = np.argmax(y_prob, axis=1)
    y_true = test_generator.classes # True labels from the generator

    # 9. Generate and Plot
    # Accuracy/Loss curves
    plt.figure(figsize=(12, 6))
    plt.plot(hist_df['accuracy'], label='Training Accuracy')
    plt.plot(hist_df['val_accuracy'], label='Validation Accuracy')
    plt.title(f'{model_name} Training and Validation Accuracy')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy')
    plt.legend()
    plt.grid(True)
    plt.show()

    plt.figure(figsize=(12, 6))
    plt.plot(hist_df['loss'], label='Training Loss')
    plt.plot(hist_df['val_loss'], label='Validation Loss')
    plt.title(f'{model_name} Training and Validation Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    plt.grid(True)
    plt.show()

    # Classification Report
    # Ensure target_names align with integer labels based on labels_map
    report = classification_report(y_true, pred_classes, target_names=list(labels_map.keys()), output_dict=True)
    report_df = pd.DataFrame(report).transpose()
    display(report_df)
    report_df.to_csv(f'{model_name}_classification_report.csv')
    print(f"Classification report saved to {model_name}_classification_report.csv")

    # Confusion Matrix
    cm = confusion_matrix(y_true, pred_classes)
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels_map.keys(), yticklabels=labels_map.keys())
    plt.title(f'{model_name} Confusion Matrix')
    plt.xlabel('Predicted Label')
    plt.ylabel('True Label')
    plt.savefig(f'{model_name}_confusion_matrix.png')
    plt.show()
    pd.DataFrame(cm, index=labels_map.keys(), columns=labels_map.keys()).to_csv(f'{model_name}_confusion_matrix.csv')
    print(f"Confusion matrix saved to {model_name}_confusion_matrix.csv")

    # ROC Curves (for multi-class classification)
    # Binarize y_true
    y_true_bin = label_binarize(y_true, classes=np.arange(NUM_CLASSES))

    # Compute ROC curve and ROC area for each class
    fpr = dict()
    tpr = dict()
    roc_auc = dict()
    for i in range(NUM_CLASSES):
        fpr[i], tpr[i], _ = roc_curve(y_true_bin[:, i], y_prob[:, i])
        roc_auc[i] = auc(fpr[i], tpr[i])

    # Compute micro-average ROC curve and ROC area
    fpr["micro"], tpr["micro"], _ = roc_curve(y_true_bin.ravel(), y_prob.ravel())
    roc_auc["micro"] = auc(fpr["micro"], tpr["micro"])

    # Compute macro-average ROC curve and ROC area
    all_fpr = np.unique(np.concatenate([fpr[i] for i in range(NUM_CLASSES)]))
    mean_tpr = np.zeros_like(all_fpr)
    for i in range(NUM_CLASSES):
        mean_tpr += np.interp(all_fpr, fpr[i], tpr[i])
    mean_tpr /= NUM_CLASSES

    fpr["macro"] = all_fpr
    tpr["macro"] = mean_tpr
    roc_auc["macro"] = auc(fpr["macro"], tpr["macro"])

    # Plot all ROC curves
    plt.figure(figsize=(10, 8))
    plt.plot(fpr["micro"], tpr["micro"],
             label=f'micro-average ROC curve (area = {roc_auc["micro"]:.2f})',
             color='deeppink', linestyle=':', linewidth=4)

    plt.plot(fpr["macro"], tpr["macro"],
             label=f'macro-average ROC curve (area = {roc_auc["macro"]:.2f})',
             color='navy', linestyle=':', linewidth=4)

    colors = cm.get_cmap('jet', NUM_CLASSES)
    for i, class_name in labels_map.items():
        plt.plot(fpr[i], tpr[i], color=colors(i), lw=2,
                 label=f'ROC curve of class {class_name} (area = {roc_auc[i]:.2f})')

    plt.plot([0, 1], [0, 1], 'k--', lw=2)
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title(f'{model_name} Multi-class ROC')
    plt.legend(loc="lower right")
    plt.savefig(f'{model_name}_roc_curve_macro_micro.png')
    plt.show()

    # 10. Store results
    models[model_name] = {
        'model': model,
        'perf': duration,
        'test_accuracy': test_accuracy,
        'test_loss': test_loss,
        'history': hist_df,
        'classification_report': report_df,
        'confusion_matrix': cm,
        'roc_auc': roc_auc
    }
    printmd(f"### {model_name} Training and Evaluation Complete.")
```
